```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F

spark = SparkSession.builder \
    .appName("Jupyter-NYC-Taxi-Distributed-Analysis") \
    .master("spark://100.111.175.49:7077") \
    .config("spark.executor.memory", "2g") \
    .config("spark.executor.cores", "4") \
    .getOrCreate()

print("✅ Connected to Spark Master!")
print("Spark Version:", spark.version)
print("Application ID:", spark.sparkContext.applicationId)
```

```python
# 2. Generate 3 Million NYC Taxi Trip Records in Spark RAM
print("Generating 3 Million NYC Taxi Trip Rows across 12 partitions...")
df_raw = spark.range(0, 3000000) \
    .withColumn("passenger_count", (F.col("id") % 6) + 1) \
    .withColumn("trip_distance", F.round((F.col("id") % 20) + 0.5, 2)) \
    .withColumn("fare_amount", F.round((F.col("id") % 50) + 2.5, 2)) \
    .withColumn("tip_amount", F.round((F.col("id") % 10) + 0.5, 2)) \
    .withColumn("total_amount", F.col("fare_amount") + F.col("tip_amount"))

df_dist = df_raw.repartition(12).cache()
print(f"Total Records: {df_dist.count():,}")
```

```python
# 3. Execute Distributed Aggregations by Passenger Count
print("Computing Aggregations by Passenger Count:")
agg_df = df_dist.groupBy("passenger_count").agg(
    F.count("*").alias("total_trips"),
    F.round(F.avg("trip_distance"), 2).alias("avg_distance_miles"),
    F.round(F.avg("fare_amount"), 2).alias("avg_fare_usd"),
    F.round(F.avg("tip_amount"), 2).alias("avg_tip_usd"),
    F.round(F.sum("total_amount"), 2).alias("total_revenue_usd")
).orderBy("passenger_count")

agg_df.show(20, truncate=False)
```

```python
# 4. Top 10 Longest Trips
print("Top 10 Longest Trips:")
top_trips = df_dist.select("id", "passenger_count", "trip_distance", "total_amount") \
    .orderBy(F.col("trip_distance").desc()) \
    .limit(10)

top_trips.show(10, truncate=False)
```

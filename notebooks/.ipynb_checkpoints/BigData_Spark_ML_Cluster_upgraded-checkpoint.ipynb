# Big Data Spark ML — Distributed Cluster Notebook

**Purpose:** Read and process a dataset (scalable up to several GB) using a real multi-node Spark cluster (`spark://spark-master:7077`), then train and compare Spark ML classification models.

This notebook was upgraded from an earlier single-model version. It keeps the original data source, cleaning logic, and `RandomForestClassifier` model, and adds:

- A fully configurable, distributed `SparkSession` (no `local[*]`)
- A cluster-verification step that proves work is running on the Workers, not the Driver
- A `TEST_MODE` switch to iterate fast on a small sample before running on the full dataset
- Richer feature engineering (`OneHotEncoder`, `StandardScaler`) inside a single `Pipeline`
- A 3-model comparison (`RandomForestClassifier`, `LogisticRegression`, and `GBTClassifier` when the label is binary)
- End-to-end benchmarking of every stage
- Partition-distribution checks and Parquet outputs

> **Settings you will most likely need to change are marked `# CHANGE ME`.**

```
Jupyter / Driver
       |
       v
  Spark Master  (spark://spark-master:7077)
       |
       +---- Worker 1
       +---- Worker 2
       +---- Worker 3
       +---- Worker 4
       +---- Worker 5
```

The **Driver** (this notebook) builds a plan and sends it to the **Master**, which schedules
tasks across the **Workers**. The Workers hold the data partitions in their executors and do
the actual computation in parallel — the Driver only collects small, aggregated results back.


## 1. Configuration

Everything you are likely to change for a different machine, cluster size, or dataset lives in this one cell.

````python
# ============================================================
# CELL — Global Configuration
# ============================================================
# All cluster/hardware/dataset knobs live here so the rest of the
# notebook never needs to be touched.

# --- Spark cluster connection --------------------------------------------
SPARK_MASTER = "spark://spark-master:7077"     # CHANGE ME if your master host/port differs

# --- Executor sizing (tune to your actual worker hardware) ----------------
EXECUTOR_MEMORY = "2g"                          # CHANGE ME to match worker RAM
EXECUTOR_CORES = 4                              # CHANGE ME to match worker CPU cores
SHUFFLE_PARTITIONS = 40                         # CHANGE ME: ~2-3x total worker cores is a good start

# --- Test / iteration mode -------------------------------------------------
# Flip TEST_MODE to False once the pipeline works end-to-end on a sample,
# to run on the full multi-GB dataset without changing anything else.
TEST_MODE = True                                # CHANGE ME: True = small sample, False = full dataset
TEST_FRACTION = 0.05                            # Fraction of the data used when TEST_MODE = True

# --- Dataset location (NO HDFS) ---------------------------------------------
# This cluster does NOT run HDFS. The dataset physically lives on the
# Spark Master's filesystem. Spark Workers/Executors run in SEPARATE
# containers and do NOT see the Master's filesystem by default — a plain
# local path here (e.g. "/data/dataset.csv") would only be readable by
# tasks that happen to run on the Master and would fail/return empty or
# partial results for tasks scheduled on the Workers.
#
# Fix: mount a SHARED Docker volume (or bind mount) at the SAME path in
# the Master container and in every Worker container, e.g. in
# docker-compose.yml:
#
#   volumes:
#     - ./shared-data:/opt/spark-data
#
# added identically to the spark-master AND every spark-worker service.
# Once every container mounts the same host folder at the same path,
# a plain local filesystem path is visible to every executor, so Spark
# can read/write it in a fully distributed way — no HDFS, no NameNode,
# no copying the dataset into the Driver/Pandas.
DATA_PATH = "/opt/spark-data/dataset.csv"        # CHANGE ME: path inside the SHARED volume (identical on Master + all Workers)
OUTPUT_PATH = "/opt/spark-data/results"          # CHANGE ME: shared-volume path for outputs
MODEL_BASE_PATH = "/opt/spark-data/models"       # CHANGE ME: shared-volume path for saved models

# --- ML column configuration (matches the original notebook's dataset) -----
LABEL_COL = "label"                             # CHANGE ME to your target column
CATEGORICAL_COLS = ["category"]                 # CHANGE ME to your categorical feature columns
NUMERIC_COLS = ["amount", "quantity"]           # CHANGE ME to your numeric feature columns

print("Configuration loaded.")
print(f"  TEST_MODE={TEST_MODE}  TEST_FRACTION={TEST_FRACTION}")
print(f"  SPARK_MASTER={SPARK_MASTER}  EXECUTOR_MEMORY={EXECUTOR_MEMORY}  EXECUTOR_CORES={EXECUTOR_CORES}")
print(f"  SHUFFLE_PARTITIONS={SHUFFLE_PARTITIONS}")
````

````python
# ============================================================
# CELL — Benchmark tracker
# ============================================================
# A tiny dict + helper used throughout the notebook to time every stage.
# This is what powers the final benchmarking summary table in Section 12.

import time

benchmark = {}
_stage_start = {}

def start_stage(name):
    _stage_start[name] = time.time()

def end_stage(name):
    benchmark[name] = time.time() - _stage_start[name]
    print(f"[benchmark] {name}: {benchmark[name]:.2f}s")

notebook_start_time = time.time()
````

## 2. Spark Session

Connects the Driver to the real Spark cluster. Deliberately **not** `local[*]` — this must schedule work on the registered Workers.

````python
# ============================================================
# CELL — Create SparkSession (connects to the real cluster)
# ============================================================
start_stage("Spark Initialization")

from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql import types as T
from pyspark import StorageLevel
from pyspark.ml import Pipeline
from pyspark.ml.feature import (
    StringIndexer,
    OneHotEncoder,
    VectorAssembler,
    StandardScaler,
)
from pyspark.ml.classification import (
    RandomForestClassifier,
    LogisticRegression,
    GBTClassifier,
)
from pyspark.ml.evaluation import MulticlassClassificationEvaluator

spark = (
    SparkSession.builder
    .appName("Jupyter-Distributed-Spark-ML")
    .master(SPARK_MASTER)
    .config("spark.executor.memory", EXECUTOR_MEMORY)
    .config("spark.executor.cores", EXECUTOR_CORES)
    .config("spark.sql.shuffle.partitions", SHUFFLE_PARTITIONS)
    .config("spark.default.parallelism", SHUFFLE_PARTITIONS)
    .config("spark.serializer", "org.apache.spark.serializer.KryoSerializer")
    .config("spark.sql.adaptive.enabled", "true")
    .config("spark.dynamicAllocation.enabled", "false")
    .getOrCreate()
)

spark.sparkContext.setLogLevel("WARN")

print("Spark version:", spark.version)
print("Spark master:", spark.sparkContext.master)
print("Application ID:", spark.sparkContext.applicationId)
print("Default parallelism:", spark.sparkContext.defaultParallelism)

end_stage("Spark Initialization")
````

## 3. Cluster Verification

This section proves the notebook is actually running on the distributed cluster and not
falling back to local execution. It sends a parallelized RDD computation through the
Master to the Workers and pulls back only the final scalar result.

```
Jupyter / Driver
       |
       v
Spark Master   <- schedules the tasks below across the Workers
       |
       +---- Worker 1   \
       +---- Worker 2    \
       +---- Worker 3     >-- each executes a slice of the 100,000-row RDD
       +---- Worker 4    /
       +---- Worker 5   /
```

Open the Spark Master UI (usually `http://<master-host>:8080`) while this cell runs —
you should see this application listed with tasks distributed across the Worker nodes.

````python
# ============================================================
# CELL — Distributed cluster test (RDD computation across Workers)
# ============================================================
sc = spark.sparkContext

test_rdd = sc.parallelize(
    range(100000),
    max(sc.defaultParallelism, 8)
)

print("Partitions:", test_rdd.getNumPartitions())
print("Default parallelism:", sc.defaultParallelism)

result = test_rdd.map(lambda x: x * x).sum()
print("Distributed test result:", result)

assert sc.master != "local[*]" and not sc.master.startswith("local"), (
    "Spark is running in local mode, not on the cluster! Check SPARK_MASTER."
)
print("Confirmed: running against a real cluster master ->", sc.master)
````

## 4. Data Loading

**No HDFS is used.** The dataset is read from a plain local filesystem path
(`DATA_PATH`) that must be mounted as a **shared Docker volume** on the Spark
Master container and on every Spark Worker container (see Section 1). The cell
below first checks, from inside the Spark cluster itself, that every Worker can
actually see `DATA_PATH` — if only the Driver/Master can see it, that's an
architecture problem and the job would silently under-process or fail on the
Workers. When `TEST_MODE = True`, a distributed `sample()` is taken *after* the
schema is inferred, so you're iterating on a representative slice of the real
data rather than a separately prepared file.

````python
# ============================================================
# CELL — Verify Workers (not just the Driver/Master) can see DATA_PATH
# ============================================================
# This runs a distributed task on the executors themselves — it does NOT
# check the path from the Driver. If DATA_PATH only exists inside the
# Master container's filesystem (and is not on a shared volume mounted
# into every Worker container too), this will fail or report False for
# some/all workers, which means the architecture problem described in
# Section 1 has not actually been fixed yet.
import os

num_probe_tasks = max(spark.sparkContext.defaultParallelism, 4)

def _check_path_visible(_):
    return os.path.exists(DATA_PATH)

worker_visibility = (
    spark.sparkContext
    .parallelize(range(num_probe_tasks), num_probe_tasks)
    .map(_check_path_visible)
    .collect()
)

print(f"DATA_PATH checked from {num_probe_tasks} distributed executor tasks:")
print("  Visible to:", sum(worker_visibility), "/", len(worker_visibility), "tasks")

if not all(worker_visibility):
    raise RuntimeError(
        f"DATA_PATH ({DATA_PATH}) is NOT visible to all Spark executors. "
        "This means the dataset is only reachable from the Master/Driver "
        "container's local filesystem, not from the Worker containers. "
        "Fix this by mounting a shared Docker volume (or bind mount) at the "
        "exact same path on the spark-master AND every spark-worker service "
        "in your docker-compose.yml — do NOT switch to HDFS and do NOT load "
        "the file into the Driver/Pandas instead."
    )
print("Confirmed: DATA_PATH is visible to every probed executor task.")
````

````python
# ============================================================
# CELL — Read dataset (distributed CSV read from shared volume, no HDFS)
# ============================================================
start_stage("Data Loading")

print("DATA_PATH:", DATA_PATH)

df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(DATA_PATH)
)

print("Columns:", df.columns)
print("Initial partitions:", df.rdd.getNumPartitions())
df.printSchema()

if TEST_MODE:
    # Distributed sample — every executor samples its own partitions,
    # nothing is collected to the Driver.
    df = df.sample(withReplacement=False, fraction=TEST_FRACTION, seed=42)
    print(f"TEST_MODE is ON — sampled ~{TEST_FRACTION:.0%} of the data for a fast iteration run.")
else:
    print("TEST_MODE is OFF — processing the full dataset.")

# Repartition once, right after loading, so every downstream stage
# (cleaning, feature engineering, training) benefits from balanced parallelism.
# We do this exactly once here rather than after every transformation, since
# repeated repartitioning just adds unnecessary shuffles.
df = df.repartition(SHUFFLE_PARTITIONS)
print("Partitions after repartition:", df.rdd.getNumPartitions())

end_stage("Data Loading")
````

## 5. Data Inspection

Inspection only ever uses `show()`/`describe()` — never `collect()` or `toPandas()` on the full dataset.

````python
# ============================================================
# CELL — Safe data inspection
# ============================================================
df.show(10, truncate=False)
print("Number of columns:", len(df.columns))

# describe() is a distributed aggregation; only the tiny summary table
# (a handful of numbers per column) ever reaches the Driver.
df.select(NUMERIC_COLS).describe().show()
````

## 6. Data Cleaning

Preserves the original duplicate-removal and numeric null-handling, and adds categorical null-handling, invalid-value filtering, and a persisted, materialized DataFrame for reuse in the following sections.

````python
# ============================================================
# CELL — Data cleaning (duplicates, nulls, invalid values, dtypes)
# ============================================================
start_stage("Preprocessing")

# Remove duplicate rows (unchanged from the original notebook)
df = df.dropDuplicates()

# Detect numeric columns present in the schema
numeric_cols_detected = [
    field.name
    for field in df.schema.fields
    if field.dataType.typeName() in ["integer", "long", "double", "float"]
]
print("Numeric columns detected:", numeric_cols_detected)

# Replace NULL numeric values with 0 (unchanged from the original notebook)
for c in numeric_cols_detected:
    df = df.withColumn(c, F.coalesce(F.col(c), F.lit(0)))

# NEW: replace NULL categorical values with an explicit "unknown" bucket
# instead of silently dropping rows or letting StringIndexer choke on nulls.
for c in CATEGORICAL_COLS:
    df = df.withColumn(c, F.coalesce(F.col(c), F.lit("unknown")))

# NEW: drop rows with an invalid (null) label — a row Spark ML cannot learn from.
df = df.filter(F.col(LABEL_COL).isNotNull())

# NEW: guard against negative values in numeric columns that should be
# non-negative (e.g. amount/quantity). Adjust this list if some numeric
# columns are legitimately allowed to be negative.
for c in NUMERIC_COLS:
    if c in df.columns:
        df = df.withColumn(c, F.when(F.col(c) < 0, F.lit(0)).otherwise(F.col(c)))

print("Cleaning completed.")

# Persist because this DataFrame is reused for: row counts, the train/test
# split, and feature engineering below. MEMORY_AND_DISK avoids OOM if the
# full (non-TEST_MODE) dataset doesn't fully fit in executor memory.
df = df.persist(StorageLevel.MEMORY_AND_DISK)

row_count = df.count()  # triggers the distributed materialization
print("Rows after cleaning:", row_count)
print("Partitions:", df.rdd.getNumPartitions())

end_stage("Preprocessing")
````

## 7. Feature Engineering

Preserves the original categorical encoding via `StringIndexer`, and adds `OneHotEncoder`
(so categories aren't treated as ordinal) plus `StandardScaler` on the numeric block. All
stages are Spark ML transformers/estimators — no Python loops over rows.

````python
# ============================================================
# CELL — Validate ML columns exist
# ============================================================
missing_cols = [
    c for c in [LABEL_COL] + CATEGORICAL_COLS + NUMERIC_COLS
    if c not in df.columns
]

if missing_cols:
    raise ValueError(
        f"These required columns are missing: {missing_cols}\n"
        f"Available columns: {df.columns}"
    )

print("All required ML columns exist.")
print("Label:", LABEL_COL)
print("Categorical:", CATEGORICAL_COLS)
print("Numeric:", NUMERIC_COLS)
````

````python
# ============================================================
# CELL — Feature engineering stages (shared by every model pipeline)
# ============================================================
start_stage("Feature Engineering")

# Label indexing (unchanged concept from the original notebook)
label_indexer = StringIndexer(
    inputCol=LABEL_COL,
    outputCol="label_index",
    handleInvalid="keep"
)

# Categorical: index -> one-hot encode (NEW: one-hot avoids implying a false
# ordinal relationship between category codes, which plain StringIndexer output does).
categorical_indexers = [
    StringIndexer(inputCol=c, outputCol=f"{c}_index", handleInvalid="keep")
    for c in CATEGORICAL_COLS
]
categorical_encoder = OneHotEncoder(
    inputCols=[f"{c}_index" for c in CATEGORICAL_COLS],
    outputCols=[f"{c}_ohe" for c in CATEGORICAL_COLS],
)

# Numeric: assemble -> scale (NEW: StandardScaler puts numeric features on a
# comparable scale, which matters for LogisticRegression in particular).
numeric_assembler = VectorAssembler(
    inputCols=NUMERIC_COLS,
    outputCol="numeric_features_raw",
    handleInvalid="keep",
)
numeric_scaler = StandardScaler(
    inputCol="numeric_features_raw",
    outputCol="numeric_features_scaled",
    withMean=True,
    withStd=True,
)

# Final feature vector: scaled numerics + one-hot categoricals
final_assembler = VectorAssembler(
    inputCols=["numeric_features_scaled"] + [f"{c}_ohe" for c in CATEGORICAL_COLS],
    outputCol="features",
    handleInvalid="keep",
)

preprocessing_stages = (
    categorical_indexers
    + [categorical_encoder, numeric_assembler, numeric_scaler, final_assembler, label_indexer]
)

print(f"Feature engineering stages ready ({len(preprocessing_stages)} stages).")

end_stage("Feature Engineering")
````

## 8. Train / Test Split

A distributed, seeded split. Both sides are persisted since each is reused across every model in the comparison.

````python
# ============================================================
# CELL — Train / test split
# ============================================================
train_df, test_df = df.randomSplit([0.8, 0.2], seed=42)

# Persisted because every model pipeline below both fits on train_df and
# scores against test_df — without this, Spark would recompute the full
# cleaning + split lineage once per model.
train_df = train_df.persist(StorageLevel.MEMORY_AND_DISK)
test_df = test_df.persist(StorageLevel.MEMORY_AND_DISK)

print("Training rows:", train_df.count())
print("Testing rows:", test_df.count())

# Decide, from the data itself, whether GBTClassifier is even applicable —
# Spark ML's GBTClassifier only supports binary classification. We check
# this rather than assuming, so we never silently misuse a model.
num_classes = train_df.select(LABEL_COL).distinct().count()
is_binary = (num_classes == 2)
print(f"Distinct classes in label column: {num_classes} (binary={is_binary})")
````

## 9. Model Training

Keeps the original `RandomForestClassifier` (same hyperparameters as the base notebook)
and adds `LogisticRegression` as a second, cheaper baseline. `GBTClassifier` is added only
if the label is binary, since Spark ML's implementation doesn't support multiclass —
adding it unconditionally would silently break on non-binary data.

````python
# ============================================================
# CELL — Build one full Pipeline per candidate model
# ============================================================
start_stage("Training")

models = {
    "RandomForestClassifier": RandomForestClassifier(
        labelCol="label_index",
        featuresCol="features",
        numTrees=50,      # CHANGE ME: more trees = better/slower, matches original notebook default
        maxDepth=10,      # CHANGE ME
        seed=42,
    ),
    "LogisticRegression": LogisticRegression(
        labelCol="label_index",
        featuresCol="features",
        maxIter=50,       # CHANGE ME
    ),
}

if is_binary:
    models["GBTClassifier"] = GBTClassifier(
        labelCol="label_index",
        featuresCol="features",
        maxIter=50,       # CHANGE ME
        maxDepth=5,       # CHANGE ME
        seed=42,
    )

fitted_models = {}
training_times = {}

for name, estimator in models.items():
    pipeline = Pipeline(stages=preprocessing_stages + [estimator])
    t0 = time.time()
    fitted_models[name] = pipeline.fit(train_df)
    training_times[name] = time.time() - t0
    print(f"Trained {name} in {training_times[name]:.2f}s")

end_stage("Training")
````

## 10. Model Evaluation

Same metric family as the original notebook (accuracy), extended with precision, recall, and F1 for a fuller picture — all computed via Spark's distributed evaluators.

````python
# ============================================================
# CELL — Score every model and collect metrics
# ============================================================
start_stage("Prediction")

predictions_by_model = {
    name: fitted.transform(test_df)
    for name, fitted in fitted_models.items()
}

end_stage("Prediction")

start_stage("Evaluation")

metric_names = ["accuracy", "weightedPrecision", "weightedRecall", "f1"]
results = []

for name, predictions in predictions_by_model.items():
    row = {"model": name, "training_time_sec": round(training_times[name], 2)}
    for metric in metric_names:
        evaluator = MulticlassClassificationEvaluator(
            labelCol="label_index",
            predictionCol="prediction",
            metricName=metric,
        )
        row[metric] = round(evaluator.evaluate(predictions), 4)
    results.append(row)
    print(f"{name}: {row}")

end_stage("Evaluation")
````

````python
# ============================================================
# CELL — Preview predictions from the best-performing model so far
# ============================================================
best_model_name = max(results, key=lambda r: r["accuracy"])["model"]
print("Best model by accuracy:", best_model_name)

predictions_by_model[best_model_name].select(
    LABEL_COL, "prediction", "probability"
).show(20, truncate=False)
````

## 11. Model Comparison Table

A small, human-readable summary — only a handful of aggregate numbers ever reach the Driver here, never raw records.

````python
# ============================================================
# CELL — Model comparison table
# ============================================================
header = f'{"Model":<24}{"Accuracy":>10}{"Precision":>12}{"Recall":>10}{"F1":>8}{"Train (s)":>12}'
print(header)
print("-" * len(header))
for row in sorted(results, key=lambda r: r["accuracy"], reverse=True):
    print(
        f'{row["model"]:<24}'
        f'{row["accuracy"]:>10.4f}'
        f'{row["weightedPrecision"]:>12.4f}'
        f'{row["weightedRecall"]:>10.4f}'
        f'{row["f1"]:>8.4f}'
        f'{row["training_time_sec"]:>12.2f}'
    )
````

## 12. Benchmarking

End-to-end timing for every stage tracked with `start_stage`/`end_stage` earlier in the notebook, so you can see exactly where time goes when you switch from `TEST_MODE=True` to the full dataset.

````python
# ============================================================
# CELL — Benchmark summary table
# ============================================================
benchmark["Total"] = time.time() - notebook_start_time

stage_order = [
    "Spark Initialization",
    "Data Loading",
    "Preprocessing",
    "Feature Engineering",
    "Training",
    "Prediction",
    "Evaluation",
    "Total",
]

header = f'{"Stage":<24}{"Time (s)":>12}'
print(header)
print("-" * len(header))
for stage in stage_order:
    if stage in benchmark:
        print(f'{stage:<24}{benchmark[stage]:>12.2f}')
````

## 13. Partition Distribution Verification

Confirms the work is actually spread across partitions (and therefore across Workers)
rather than sitting in one giant partition. Only the small per-partition counts —
never the underlying rows — are collected to the Driver.

````python
# ============================================================
# CELL — Partition distribution check
# ============================================================
print("Spark Master:", spark.sparkContext.master)
print("Default parallelism:", spark.sparkContext.defaultParallelism)
print("Number of partitions (df):", df.rdd.getNumPartitions())

# Count rows per partition without collecting any actual row data —
# mapPartitions returns one small integer per partition.
partition_sizes = (
    df.rdd
    .mapPartitions(lambda it: [sum(1 for _ in it)])
    .collect()
)

print("Rows per partition:", partition_sizes)
print("Min / Max / Avg partition size:",
      min(partition_sizes), max(partition_sizes),
      round(sum(partition_sizes) / len(partition_sizes), 1))

print("\nTo verify Workers are actually processing this job, open the Spark")
print("Master UI (typically http://<master-host>:8080) and confirm:")
print("  - Multiple Workers are ALIVE")
print("  - This application has executors running on more than one Worker")
print("  - The Stages tab shows tasks distributed across those executors")
````

## 14. Save Results

Writes predictions and the best model to Parquet/Spark ML format, as in the original notebook, but for every model plus a clearly-marked best-model export.

````python
# ============================================================
# CELL — Save predictions (Parquet, distributed write)
# ============================================================
for name, predictions in predictions_by_model.items():
    path = f"{OUTPUT_PATH}/{name}"
    (
        predictions
        .select(LABEL_COL, "prediction", "probability")
        .write
        .mode("overwrite")
        .parquet(path)
    )
    print(f"Saved predictions for {name} -> {path}")
````

````python
# ============================================================
# CELL — Save the best model
# ============================================================
best_model_path = f"{MODEL_BASE_PATH}/{best_model_name}"
fitted_models[best_model_name].write().overwrite().save(best_model_path)
print(f"Best model ({best_model_name}) saved to: {best_model_path}")
````

## 15. Final Distributed Execution Verification

````python
# ============================================================
# CELL — Final cluster check
# ============================================================
print("Application ID:", spark.sparkContext.applicationId)
print("Master:", spark.sparkContext.master)
print("Default Parallelism:", spark.sparkContext.defaultParallelism)
print("Final partitions (df):", df.rdd.getNumPartitions())

print("\nCheck the Spark Master UI for:")
print("- Registered Workers")
print("- Active Executors")
print("- Running Jobs")
print("- Stages / Tasks")

print("\nTEST_MODE was:", TEST_MODE, "| TEST_FRACTION was:", TEST_FRACTION)
````
